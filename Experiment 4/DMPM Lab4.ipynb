```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load the dataset
df = pd.read_csv("https://raw.githubusercontent.com/mwaskom/seaborn-data/master/diamonds.csv")

# =========================================================
# 1. Box Plot: Compare Price distribution by Cut
# =========================================================
sns.boxplot(data=df, x="cut", y="price")
plt.title("1. Box Plot: Price by Cut")
plt.xlabel("Cut")
plt.ylabel("Price ($)")
plt.show()

# =========================================================
# 2. Scatter Plot: Relationship between Carat & Price
# =========================================================
sns.scatterplot(data=df.sample(500, random_state=1), x="carat", y="price", hue="cut")
plt.title("2. Scatter Plot: Carat vs. Price")
plt.xlabel("Carat")
plt.ylabel("Price ($)")
plt.show()

# =========================================================
# 3. Histogram: Distribution of Carat
# =========================================================
sns.histplot(df["carat"])
plt.title("3. Histogram: Distribution of Carat")
plt.xlabel("Carat")
plt.ylabel("Count")
plt.show()

# =========================================================
# 4. Line Chart: Trend of Price across samples
# =========================================================
plt.plot(df["price"].head(30), label="Price")
plt.title("4. Line Chart: Price for First 30 Samples")
plt.xlabel("Sample Index")
plt.ylabel("Price ($)")
plt.grid(True)
plt.show()

# =========================================================
# 5. Pie Chart: Percentage breakdown of Cut
# =========================================================
cut_counts = df["cut"].value_counts()
plt.pie(cut_counts, labels=cut_counts.index, autopct="%1.1f%%")
plt.title("5. Pie Chart: Distribution of Cut Quality")
plt.show()

# =========================================================
# 6. Bar Plot: Average Price by Cut
# =========================================================
mean_price = df.groupby("cut")["price"].mean().reset_index()
sns.barplot(data=mean_price, x="cut", y="price")
plt.title("6. Bar Plot: Average Price by Cut")
plt.xlabel("Cut")
plt.ylabel("Average Price ($)")
plt.show()
```
