**wap to find the mean of the datapaints without using functions**

```python
datapoints = [15, 13, 18, 16, 14,12,17]
total = sum(datapoints)
count = len(datapoints)
mean = total / count
print("Mean:", mean)
```

**wap to print the average mean of data points with using functions**

```python
from statistics import mean
data_points = [12, 15, 22, 8, 19, 25, 14]
print("Average (Mean):", mean(data_points))
```

**wap to find median of the data points without using function**

```python
data = [12, 15, 22, 8, 19, 25, 14]
data.sort()
n = len(data)
median = data[n//2] if n % 2 == 1 else (data[n//2 + 1] + data[n//2]) / 2
print("Median:", median)
```

```python
data = [12, 15, 22, 8, 19, 25, 14, 15]
data.sort()
n = len(data)
if n % 2 == 1:
    median = data[n // 2]
else:
    median = (data[n // 2 - 1] + data[n // 2]) / 2

print("Median:", median)
```

**wap to find the mode of a data items without using functions**

```python
data = [15,13,18,16,14,12,12]
f = []
for i in data:
    if data.count(i) > 1:
       f.append(i)
print("mode:",max(f))
```

**wap to find the mode of a data items with using functions**

```python
from statistics import mode
data_points = [12,12, 15, 22, 8, 19, 25, 14]
print("Average (Mode):", mode(data_points))
```

**wap to print the varience of data points without function**

```python
data = [12, 15, 22, 8, 19, 25, 14]
mean = sum(data) / len(data)
variance = sum((x - mean) ** 2 for x in data) / len(data)
print("Variance:", variance)
```

**wap to print the varience of data points without function**

```python
from statistics import variance
data_points = [12, 15, 22, 8, 19, 25, 14,12]
print("Average (Variance):", variance(data_points))
```

**wap to print the standard daviation with and without using function**

```python
datapoints = [12, 15, 22, 8, 19, 25, 14]
mean = sum(datapoints) / len(datapoints)
variance = sum((x - mean) ** 2 for x in datapoints) / len(datapoints)
std_deviation = variance ** 0.5
print("Standard Deviation:", std_deviation)
```

**wap to find the varience and standard daviation of a datapoints entered by the user**

```python
datapoints_str = input("enter the datapoints comma separated :")
datapoints = datapoints_str.split(",")
datapoints = [int(x) for x in datapoints]
mean = sum(datapoints) / len(datapoints)
variance = sum((x - mean) ** 2 for x in datapoints) / len(datapoints)
std_deviation = variance ** 0.5
print("variance",variance)
print("standard deviation",std_deviation)
```
